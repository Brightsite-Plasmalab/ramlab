# Wigner $3j$ symbols
Validation against analytical expressions from Satija (2020)

```python
from ramlab.molecules.common import wigner_3j
import numpy as np
import matplotlib.pyplot as plt
import wedme.apply.dev
```

```python
J = np.arange(3/2, 100/2, 1)
```

```python
# Satija (2020) eq. 62
wigner3j_sim = wigner_3j(J, J, 2, 1/2, -1/2, 0)
wigner3j_calc = (-1) ** (J - 1/2) * (3/4 - J * (J + 1)) / np.sqrt((2*J + 3) * (J + 1) * (2* J+1) * (J * (2*J - 1)))

plt.figure()
plt.plot(J, wigner3j_sim, '.', label='sim')
plt.plot(J, wigner3j_calc, 'o', label='calc', markerfacecolor='none')
plt.legend()
plt.xlabel('J')
plt.ylabel('wigner3j')
plt.show()
```

```python
# Satija (2020) eq. 67
wigner3j_sim = wigner_3j(J, J-1, 2, 3/2, -3/2, 0)
wigner3j_calc = 3/4 * (-1) ** (J + 1/2) * \
    np.sqrt(3*(2*J+3)*(2*J-3)) \
        / np.sqrt((J+1) * (2*J+1) * J * (2*J-1) * (J-1))

plt.figure()
plt.plot(J, wigner3j_sim, '.', label='sim')
plt.plot(J, wigner3j_calc, 'o', label='calc', markerfacecolor='none')
plt.legend()
plt.xlabel('J')
plt.ylabel('wigner3j')
plt.show()
```
